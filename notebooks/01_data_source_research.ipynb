```python
import requests
from bs4 import BeautifulSoup

url = "https://www.24h-rennen.de/en/live-en/"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print("STATUS CODE:")
print(response.status_code)

print("\nFIRST 1000 CHARACTERS:\n")
print(response.text[:1000])
```

```python
keywords = [
    "api",
    "json",
    "socket",
    "websocket",
    "live",
    "timing",
    "graphql"
]

html = response.text.lower()

print("KEYWORD SEARCH RESULTS:\n")

for keyword in keywords:
    if keyword in html:
        print(f"FOUND: {keyword}")
```

```python
soup = BeautifulSoup(response.text, "html.parser")

scripts = soup.find_all("script")

print(f"TOTAL SCRIPTS FOUND: {len(scripts)}\n")

for script in scripts:
    src = script.get("src")
    
    if src:
        print(src)
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
import time

# Setup Chrome
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install())
)

# Open live page
url = "https://www.24h-rennen.de/en/live-en/"
driver.get(url)

# Wait for page to load
time.sleep(10)

# Print page title
print("PAGE TITLE:")
print(driver.title)

# Print first 1000 chars
print("\nPAGE SOURCE PREVIEW:\n")
print(driver.page_source[:1000])

# Keep browser open for inspection
```

```python
from selenium.webdriver.common.by import By

# Find all tables
tables = driver.find_elements(By.TAG_NAME, "table")

print(f"TOTAL TABLES FOUND: {len(tables)}")
```

```python
table_html = tables[0].get_attribute("outerHTML")

print(table_html[:3000])
```

```python
import pandas as pd

df = pd.read_html(table_html)[0]

df.head()
```

```python
for i, table in enumerate(tables):
    print(f"\nTABLE INDEX: {i}")
    print("-" * 50)
    print(table.text[:500])
    print("\n" + "="*70)
```

```python
iframes = driver.find_elements(By.TAG_NAME, "iframe")

print(f"TOTAL IFRAMES FOUND: {len(iframes)}")

for i, iframe in enumerate(iframes):
    print(f"\nIFRAME INDEX: {i}")
    print("SRC:", iframe.get_attribute("src"))
```

```python
# Switch into live timing iframe
driver.switch_to.frame(iframes[2])

print("Switched into iframe successfully")
```

```python
import os
from datetime import datetime

os.makedirs("../data/raw", exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

raw_path = f"../data/raw/live_raw_{timestamp}.csv"

df.to_csv(raw_path, index=False)

print(f"RAW DATA SAVED: {raw_path}")
```

```python
clean_df = df.copy()
```

```python
clean_df = clean_df.drop(columns=["Unnamed: 0"])
```

```python
print(clean_df.isnull().sum())
```

```python
core_columns = [
    "Pos.",
    "No.",
    "Class",
    "Pro",
    "Rank",
    "Name",
    "Laps",
    "Gap",
    "Last",
    "Fastest",
    "Pit",
    "Vehicle"
]

race_df = clean_df[core_columns]

race_df.head()
```

```python
os.makedirs("../data/processed", exist_ok=True)

processed_path = f"../data/processed/clean_race_data_{timestamp}.csv"

race_df.to_csv(processed_path, index=False)

print(f"CLEAN DATA SAVED: {processed_path}")
```
